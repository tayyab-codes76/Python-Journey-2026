# ***🧩 Problem 1: Student Data System***

>### 🎯 Task:

>1.Print all student names\
>2.Print skills of 2nd student\
>3.Print average marks of each student\
>4.Collect all unique skills using a set

```python
student = [
    {"name":"Ali", "skills":["python","ML"],"marks":(80,90)},
    {"name":"Ahmed","skills":["python","web"],"marks":(70,85)},
    {"name":"Sara","skills":["ML","Data"],"marks":(88,92)}
]

# 1.All students name
for s in student:
    print(s["name"])
# 2.Skills of 2nd student
print("These are the skills of second Student",student[1]["skills"])
# 3.Average marks of all students
for s in student:
    avg = sum(s["marks"])/len(s["marks"])
    print(s["name"], avg)
# 4.Collect all unique skills using set
unique_skills = set()
for s in student:
    unique_skills.update(s["skills"])
print(unique_skills)

```

# 🧩 Problem 2: Shopping System

>1.Remove duplicates using set\
>2.Print total cost of cart\
>3.Print each item with its price

```python
cart = ["apple", "banana", "apple", "orange", "banana"]
prices = {
    "apple": 50,
    "banana": 30,
    "orange": 40
}

#1.Remove duplicates using set
cart = set(cart)
print("All unique items: ",cart)

#2.Print total cost of cart
value = prices.values()
total_cost = sum(value)
print("The total cost is: ",total_cost)

#3.Print each item with its price
for key,value in prices.items():
    print("Keys and values: ",key,value)
    
```

```python
cart = ["apple", "banana", "apple", "orange", "banana"]
prices = {
    "apple": 50,
    "banana": 30,
    "orange": 40
}

# 1. Unique items
unique_items = set(cart)
print("Unique items:", unique_items)

# 2. Total cost
total_cost = 0
for item in cart:
    total_cost += prices[item]
print("Total cost:", total_cost)

# 3. Each item with price
for item in cart:
    print(item, prices[item])
```
