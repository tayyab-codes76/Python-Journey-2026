# ***Dictionary***

>### Creating Dcitionary

```python
empty = {}
person_dict = {"name": "Tayyab", "age":20,"city":"Bhalwal","class":"Bs Data science"}
# Accessing value
print(person_dict["name"])
print(person_dict["city"])
#adding and updating value
person_dict["age"] = 21 #Updating value
person_dict["grade"] = "A" # Adding value
print(person_dict)
```

>### Removing elements



```python
person_dict = {"name": "Tayyab", "age":20,"city":"Bhalwal","class":"Bs Data science"}
# Removing element
person_dict.pop("name")
print(person_dict)
del person_dict["class"]#delet
print(person_dict)
# Empty dictionary
person_dict.clear()
print(person_dict)
```

>### Looping with dictionary

```python
person_dict = {"name": "Tayyab", "age":20,"city":"Bhalwal","class":"Bs Data science"}
#keys
for key in person_dict:
    print(key)
#Value
for value in person_dict.values():
    print(value)
    # key+value
for key, value in person_dict.items():
        print(key,":",value)
```
