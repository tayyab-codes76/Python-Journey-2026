# ***For Loop***

>### Iterating a list of a number

```python
num = [1,2,3,4,5,6]
for n in num:
    print(n)
```

>### Iterating a list of a numbers with there indexing values

```python
# print all elements whith there indexing
num = [1,2,3,5,6]
index = 0
for n in num:
    print(f"{index}: {n}")
    index += 1

# using enumerate function
for index, value in enumerate(num):
    print(index, value)

for index, value in enumerate(num, start = 3):# you can indexing where you want
    print(index, value)
```

>### Iterating a list of fruits with there indexing numbers

```python
# Using enumerate function
fruites = ["apple", "banana", "cherry", "mango", "orange"]
for index, value in enumerate(fruites, start=1):
    print(index,":", value)

# manually
index = 0
for fruite in fruites:
    print(f"{index}: {fruite}")
    index += 1
```

>### Iterrating on dictionary with for loop

```python
# Iterating over dictionaries
my_self = {
    "name":"Tayyab",
    "age":20,
    "city":"Bhawal",
    "hobby":"chess"
    
}
# print just keys.
print("These are just keys:")
for key in my_self:
    print(key)

# print just values in dictionary
print("These are just values:")
for value in my_self.values():
    print(value)

# key value pairs
print("These are both keys and values:")
for key, value in my_self.items():
    print(key,value)
```

>### 2.Sum of Numbers

* Ask the user for a number n.
* Print the sum of all numbers from 1 to n.

```python
n = int(input("Enter a numer bro: "))
total = 0
for i in range(1,n):
    total += i
print(total)
```
