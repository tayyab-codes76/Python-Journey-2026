>#                                 Accessing elements

```python
# Accessing elements with indexing
fruits = ["apple", "banana", "mango", "oring"]
print(fruits)
print(fruits[0])
print(fruits[3])
print(fruits[-1])
```

>#                        Modifyng elements

```python
fruits = ["apple", "banana", "mango", "oring"]
print(fruits)
fruits[0] = "mango"
print(fruits)
```

>#                   Adding elements

```python
fruits = ["apple", "banana", "mango", "oring"]
fruits.append("cherry") # adding the element at the end of list.
fruits.insert(0,"grasp" ) # adding the element by accessing indexing point.
fruits.extend(["grap", "kiwi"]) # adding multiple element at the end of the list.
print(fruits)
```

>#                        Removing elements

```python
fruits = ["apple", "banana", "mango", "oring"]
fruits.pop() # removing the element at the end of the list.
fruits.remove("banana") # removing by value.
fruits.pop(0) # removing by indexing.
print(fruits)
```

>#                          List operations

```python
a = [1,2,3]
b = [4,5]
print(a+b) # Concatination
print(a*5)# repetition
print(3 in a) # membership checking
print(10 not in a) # membership checking
```

>#                              Slicing with list

```python
nums = [1,2,3,4,5,6,7]
print(nums[2:5]) # start to index 2 to indx 4
print(nums[:3])  # start to index 2
print(nums[3:])   # index 3 to end
print(nums[-4:-1])# negative slicing
print(nums[::-1]) # reverse list
  
```

>#                         Nested list

```python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

print(matrix[0][1])
matrix[2][2] = 99
print(matrix)
```

>#                        print elements of list with indexing numbers

```python
fruits = ["apple", "banana", "mango", "cherry", "graps"]
for index, value in enumerate(fruits, start=1):
    print(index,value)
```
