># ***Beginner Level (Simple Conditions)***

>### 1️⃣ Traffic Light System (Basic) Write a program that asks for a traffic light color (Green, Yellow, Red) and prints the correct action:

>1.Green → "You can GO!"\
>2.Yellow → "Slow down and prepare to STOP!"\
>3.Red → "STOP!"

```python
traffic_light = input("Enter the traffic light color. Please write color name in small characters.")
if traffic_light == "green":
    print("You can Go!")
elif traffic_light == "yellow":
    print("Slow down and prepare to STOP!")
elif traffic_light == "red":
    print("STOP!")
else:
    print("right correct color name!!!")
```

>### 3️⃣ Categorizing Temperature Levels Write a program that takes a temperature (°C) and prints whether it's:

>1.Above 30°C → "It's hot outside!"\
>2.Between 15°C and 30°C → "It's pleasant weather."\
>3.Below 15°C → "It's cold outside!"

```python
temperature = int(input("Enter the tempratur: "))
if temperature > 30:
    print("It's hot outside!")
elif temperature >= 15 and temperature <= 30:
    print("It's pleasant weather!")
else:
    print("It's cold outside!")
```

>### 4️⃣ Exam Pass or Fail Write a program that asks for a student’s exam score (0-100) and assigns a grade:

>* 90-100 → "Grade A"
>* 80-89 → "Grade B"
>* 70-79 → "Grade C"
>* 60-69 → "Grade D"
>* Below 60 → "Grade F"

```python
marks = int(input("Enter you are marks bro!"))
if marks >= 90 and marks <= 100:
    print("Grad A")
elif marks >= 80 and marks <= 89:
    print("Grad B")    
elif marks >= 70 and marks <= 79:
    print("Grad C")
elif marks >= 60 and marks <= 69:
    print("Grad D")
else:
    print("Grad F")
```

>### 5️⃣ Checking Leap Year Write a program that asks for a year and checks whether it is a leap year or not.

>##### Hint: A year is a leap year if:

* It is divisible by 4
* But not divisible by 100, unless also divisible by 400

```python
year = int(input("Enter the year: "))
if (year%4 == 0 and year%100 != 0) or (year%400 == 0):
    print("It's a leap year.")
else:
    print("It's not a leap year.")
```
