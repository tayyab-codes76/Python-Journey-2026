# ***Set***

>### Creating sets

```python
empty = set() # empty set (cannot use{} that is a dict)
numbers = {2,3,5,23,85}
fruits = { "apple", "banana", "orange"}
mixed = {1, "apple", True, 3.14}
```

>### Adding Elements

```python
fruits = { "apple", "banana", "orange"}
# add single elements
fruits.add("mango")
# add multiple elements
fruits.update(["mango", "grap"])
print(fruits)
```

>### Removing element

```python
fruits = { "apple", "banana", "orange"}
# removes element, error if not present
fruits.remove("banana")
# removes element, no error if not present
fruits.discard("kiwi")
# removed rendom element
popped = fruits.pop()
# empty the set
fruits.clear()
print(fruits)
```

>### Set Operations (Mathematical style)

```python
A = {1,2,3,4}
B = {3,4,5,6}
# Union
print(A|B)
# Intersection
print(A&B)
# Difference
print(A - B)
# Symmetric difference
print(A^B)
```

>### Quick Practice Challenge

>1.Create a set of 5 numbers → print it\
>2.Add 2 more numbers → print it\
>3.Remove 1 number → print it\
>4.Create another set → find union, intersection, difference, symmetric difference\
>5.Check if a number exists in the set

```python
# 1.Create a set of 5 numbers → print it
num = {1,2,3,4,5}
print(num)
# 2.Add 2 more numbers → print it
num.update([6,7])
print(num)
# 3.Remove 1 number → print it
num.remove(7)
print(num)
# 4.Create another set → find union, intersection, difference, symmetric difference
num = {1,2,3,4,5}
num1 = {8,3,4,6,8,2,3}
# Union
print(num|num1)
# Intesection
print(num&num1)
# difference
print(num - num1)
# symmetric difference
print(num ^ num1)
```
