# ***Tuple***

>### Creating Tuples

```python
empty = ()
single = (5,) # single element of a tuple must have a comma
numbers = (1,2,3,4,5)
fruits = ("apple", "banana", "oring", "mango", "cherry")
mixed = (1, "apple", True, 3.13)
```

>### Accessing elements

```python
fruits = ("apple", "banana", "oring", "mango", "cherry")
fruits[0]
fruits[1:3]
fruits[::-1]
```

>### Tuple Operations

```python
a = (1,2,3)
b = (4,5)
print(a+b) # concatinations
print(a*3) # repetitions
print(2 in a) # checking membership
print(len(a))# length
```

>### Accessing elements by indexing 

```python
fruits = ("apple", "banana", "oring", "mango", "cherry")
for index, value in enumerate(fruits, start = 1):
    print(index,value)
```

>### Tuple unpacking trick

```python
point = (10, 20, 30)
x,y,z = point
print(x,y,z)

```
