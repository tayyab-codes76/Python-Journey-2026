># 🔹 Basic Tasks

>### 1.Print number 1 to 20

```python
i = 1
while i <= 20:
    print(i)
    i += 1
```

>### 2.print even numbers till 50

```python
number = 2
while number <= 50:
    print(number)
    number += 2
```

>### 3.print table of 5

```python
n = 5
i= 1
while i <= 10:
    print(f"{n}x{i} = {n*i}")
    i += 1
```

>### 4. Sum of numbers from 1 to n

```python
total = 0
number = int(input("Enter a number: "))
i = 1
while i <= number:
    total += i
    i += 1
    
print(total)
```
