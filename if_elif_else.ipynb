# ***If_elif_else conditions***

```python
# if condition
traffic_light = "green"
if traffic_light == "green":
    print("You can go")

# else condition
traffic_light = "red"
if traffic_light == "green":
    print("You can go!")
else:
    print("you must stop. Because, the signal is not green.")

#elif condition
traffic_light = "red"
if traffic_light == "green":
    print("you can go!")
elif traffic_light == "red":
    print("you must stop!")
else:
    print("slow down and prepare for stop!")
```

```python
# Define traffic light color
traffic_light = "yellow"
# Define the road is clear
road_clear = False
# Checking the traffic conditions
if traffic_light == "green":
    # Nested condition check if the road is clear
    if road_clear:
        print("You can cross road safely")
    else:
        print("wait! The road is not clear. Even though the light is green.")
elif traffic_light == "yellow":
    print("Slow down and prepare for srop.")
else:
    print("You must stop!")
```

>### 3. Grade System

>* Take marks (0–100)\
Print grade:

* 90+ → A
* 80–89 → B
* 70–79 → C
* 60–69 → D
* Below 60 → F

```python
marks = int(input("Enter you marks Sir."))
if marks >= 90 and marks <= 100:
    print("A")
elif marks >= 80 and marks <= 89:
    print("B")
elif marks >= 70 and marks <= 79:
    print("C")
elif marks >= 60 and marks <= 69:
    print("D")
else:
    print("F")
```

# ***🔴 Advanced Level***

>### 5. Simple Calculator
>Take:

* First number
* Second number
* Operator (+, -, *, /)
* Perform operation using if, elif, else
* Print result
* If invalid operator → print error

```python
num1 = int(input("Eter the first number: "))
num2 = int(input("Enter the second number: "))
operators = input("what you want to do: +,-,*,/")
print(f"First number is: {num1}\nSecond number is: {num2}\nThe operator is: {operators}")
if operators == "+":
    print(num1 + num2)
elif operators == "-":
    print(num1 - num2)
elif operators == "*":
    print(num1*num2)
elif operators == "/":
    if num2 == 0:
        print("Devided by 0 is not allowed!")
    else:
        print(num1/num2)
else:
    print("Error!")
```
