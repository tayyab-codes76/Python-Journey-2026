# Set operations

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
#Union: all elements from both sets
print(f"Union: {set_a.union(set_b)}")
# Intersection: Commen element from both sets
print(f"Intersection: {set_a.intersection(set_b)}")
# Difference: Element in set a but not in set b
print(f"Difference: {set_a.difference(set_b)}")
# symmetric difference: Elements in either set a or set b but not in both
print(f"Symmetric Difference: {set_a.symmetric_difference(set_b)}")
```
