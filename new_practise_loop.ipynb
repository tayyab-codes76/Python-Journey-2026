# ***Using itertools to cycle through a list***

```python
import itertools
colors = ["red", "green", "blue"]
counter = 0
print("Cycling through colors using itertools.cycle:")
for color in itertools.cycle(colors):  
    print(color)
    counter += 1
    if counter == 6:  
        break
```

# ***While Loop***

>#   Instructions.

#      ***3 things are essential in while loop.***

* Starting point
* Condition
* Update(increment/decrement)

#      ***Flow of while loop***

1. Condition check.
2. If contidion True, Code run. if condition False, code stop.
3. Update the conditon.
4. Again check condition. 

***This cycle repeated.***

```python
i = 1 # starting point
while i <= 10: # Conditiion lgai k jeb tuk i 10 say chota yay braber hay tub tuk loop chlao.
    print(i)
    i += 1 # update kia.
```

>#      Break
Break mean's stop the loop intirely.

```python
i = 1
while i <= 10:
    if i == 5:
        break
    print(i)
    i += 1
```

>#      Continue
Continue means skip this round, and go waya at next iteration.

```python
i = 1
while i <= 10:
    i += 1
    if i == 5:
        continue
    print(i)
    
```
